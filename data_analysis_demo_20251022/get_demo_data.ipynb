```python
import csv
import datetime
import efinance as ef


# 股票代码 贵州茅台
stock_code = '600519'
# 数据间隔时间为 1 分钟
freq = 1
# 获取最新一个交易日的分钟级别股票行情数据
#df = ef.stock.get_quote_history(stock_code, klt=freq)

df = ef.stock.get_quote_history(stock_code, beg='20231201', end='20231201', klt=1, fqt=1)

# 将数据存储到 csv 文件中
df.to_csv('demo_data.csv', encoding='utf-8-sig', index=None)
print(f'股票: {stock_code} 的行情数据已存储到文件: demo_data.csv 中！')

# 读取CSV文件并将日期字符串转换为Python datetime对象
with open('demo_data.csv') as f:
    reader = csv.reader(f)
    headers = next(reader)
    data = []
    for row in reader:
        date_str = row[2]
       # print(date_str)
        date_obj = datetime.datetime.strptime(date_str, '%Y-%m-%d %H:%M')
        value = float(row[10])
        data.append((date_obj, value))

# 从CSV文件中提取数据并创建列表，其中日期字符串已被转换为Python datetime对象
dates = [d[0] for d in data]
values = [d[1] for d in data]



```
